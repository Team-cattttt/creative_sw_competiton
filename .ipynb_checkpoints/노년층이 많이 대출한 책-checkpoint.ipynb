```python
import pandas as pd
df = pd.read_csv("book_60.csv", encoding = 'utf-8')
df
```

```python
df = df.drop(df.columns[2:9], axis = 1)
df
```

```python
df = df.drop(df.columns[0], axis = 1)
df
```

시리즈물 제거

```python
df = df.drop_duplicates(['서명'])
df
```

이제 상위 20개의 도서만 뽑아보겠습니다.

```python
book_rank = df.head(10)
book_rank
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
fig = px.bar(data_frame=book_rank, x="서명", y="대출건수")
fig.show()
```

```python
fig = px.bar(data_frame=book_rank, x="서명", y="대출건수", orientation='h')
fig.show()
```
