```python
import numpy as np
import matplotlib.pyplot as plt

from matplotlib import lines
from matplotlib import patches
from matplotlib.patheffects import withStroke


# 막대 그래프 생성
plt.rc('font', family='Malgun Gothic')
plt.figure(figsize=(12, 20))
bars = plt.barh(book_rank['서명'], book_rank['대출건수'], color='skyblue')
plt.xlabel('대출건수')
plt.ylabel('순위')
plt.title('노년층이 많이 대출한 책', size = 18, fontweight='bold')

plt.yticks(ticks=np.arange(len(book_rank)), labels=np.arange(1, len(book_rank) + 1), rotation=45)
plt.gca().invert_yaxis()

for i, (bar, name) in enumerate(zip(bars, book_rank['서명'])):
    if i == 9 or i == 12 or i == 18 or i == 19:  # 1번과 2번 막대의 경우
        plt.text(bar.get_width(), bar.get_y() + bar.get_height() / 2, name,
                 va='center', ha='left', color='black', fontsize=10, weight='bold')
    else:  # 나머지 막대의 경우
        plt.text(bar.get_width() / 2, bar.get_y() + bar.get_height() / 2, name,
                 va='center', ha='center', color='black', fontsize=10, weight='bold')

plt.tight_layout()

plt.savefig('노년층이 많이 대출한 책.jpg', format='jpeg')
```

```python
# 최종

import pandas as pd
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.common.exceptions import NoSuchElementException
import time
import os


# CSV 파일 읽기
output_directory = "C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회"  # 출력 CSV 파일을 저장할 디렉토리

# 출력 파일 경로 생성
output_file_name = "노년층이 많이 대출한 책.xlsx"
output_file_path = os.path.join(output_directory, output_file_name)

# 첫 번째 열의 내용 가져오기
queries = book_rank.iloc[:, 0].tolist()

# 크롬 드라이버 경로 설정 및 웹 브라우저 실행
path = "C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회/chromedriver-win64/chromedriver.exe"
service = Service(path)
driver = webdriver.Chrome(service=service)

# Yes24 웹사이트 열기
driver.get("https://www.yes24.com/main/default.aspx")
time.sleep(2)

# 검색 결과를 저장할 리스트
results = []

# 각 검색어에 대해 검색 수행
for query in queries:
    search_box = driver.find_element(By.ID, "query")
    search_box.clear()
    search_box.send_keys(query)
    
    driver.find_element(By.XPATH, '//*[@id="yesSForm"]/fieldset/span[2]/button').click()
    
    time.sleep(2)
    
    # 검색 결과 중 첫 번째 항목 클릭
    try:
        first_result = driver.find_element(By.XPATH, '//*[@id="yesSchList"]/li/div/div[2]/div[2]/a[1]')
        first_result.click()
        time.sleep(2)
        
        # 저자 정보 추출
        author = driver.find_element(By.XPATH, '//*[@id="yDetailTopWrap"]/div[2]/div[1]/span[2]/span[1]/a')
        author_text = author.text
        
        # 책소개 정보 추출
        book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_inBook"]/div[2]/div[1]/div[1]')
    except NoSuchElementException:
        try :
            book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_introduce"]/div[2]/table/tbody/tr/td/div[2]/div')
    
        except NoSuchElementException:
            try :
                book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_introduce"]/div[2]/table/tbody/tr/td/div/div')
            # 두 번째 XPath로도 요소를 찾지 못했을 경우 예외 처리
            except NoSuchElementException:
                book_intro = None
                
    if book_intro:
        book_intro_text = book_intro.text
    else :
        continue
        
        # 데이터 프레임에 추가
        results.append({
            '검색어': query,
            '저자': author_text,
            '책소개': book_intro_text
        })
        
        driver.back()
        time.sleep(2)

# 드라이버 종료
driver.quit()

df = pd.DataFrame(results)
os.makedirs(output_directory, exist_ok=True)

with pd.ExcelWriter(output_file_path, engine='openpyxl') as writer:
     df.to_excel(writer, index=False, startrow=0, startcol=0)

print(f"검색 결과가 {output_file_path}에 저장되었습니다.")
```

```python
# 최종, 하나 빼고 잘 됨.

import pandas as pd
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import NoSuchElementException
import time
import os


# CSV 파일 읽기
output_directory = "C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회"  # 출력 CSV 파일을 저장할 디렉토리

# 출력 파일 경로 생성
output_file_name = "노년층이 많이 대출한 책.xlsx"
output_file_path = os.path.join(output_directory, output_file_name)

# 첫 번째 열의 내용 가져오기
queries = book_rank.iloc[:, 0].tolist()

# 크롬 드라이버 경로 설정 및 웹 브라우저 실행
path = "C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회/chromedriver-win64/chromedriver.exe"
service = Service(path)
driver = webdriver.Chrome(service=service)

# Yes24 웹사이트 열기
driver.get("https://www.yes24.com/main/default.aspx")
time.sleep(2)

# 검색 결과를 저장할 리스트
results = []

# 각 검색어에 대해 검색 수행
for query in queries:
    try:
        search_box = driver.find_element(By.ID, "query")
        search_box.clear()
        search_box.send_keys(query)
        
        driver.find_element(By.XPATH, '//*[@id="yesSForm"]/fieldset/span[2]/button').click()
        
        time.sleep(2)
        # 검색 결과 중 첫 번째 항목 클릭
        first_result = WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable((By.XPATH, '//*[@id="yesSchList"]/li/div/div[2]/div[2]/a[1]'))
        )
        first_result.click()
        time.sleep(2)

        # 저자 정보 추출
        author = WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.XPATH, '//*[@id="yDetailTopWrap"]/div[2]/div[1]/span[2]/span[1]/a'))
        )
        author_text = author.text
        
        # 책소개 정보 추출
        try:
            book_intro = WebDriverWait(driver, 10).until(
                EC.presence_of_element_located((By.XPATH, '//*[@id="infoset_inBook"]/div[2]/div[1]/div[1]'))
            )
            book_intro_text = book_intro.text
        except NoSuchElementException:
            try:
                book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_introduce"]/div[2]/table/tbody/tr/td/div[2]/div')
                book_intro_text = book_intro.text
            except NoSuchElementException:
                try:
                    book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_introduce"]/div[2]/table/tbody/tr/td/div/div')
                    book_intro_text = book_intro.text
                except NoSuchElementException:
                    book_intro_text = None

        # 결과 저장
        results.append({
            '검색어': query,
            '저자': author_text,
            '책소개': book_intro_text
        })

        driver.back()
        time.sleep(2)
    except Exception as e:
        print(f"Error processing query '{query}': {e}")
        continue


# 드라이버 종료
driver.quit()

df = pd.DataFrame(results)
os.makedirs(output_directory, exist_ok=True)

with pd.ExcelWriter(output_file_path, engine='openpyxl') as writer:
     df.to_excel(writer, index=False, startrow=0, startcol=0)

print(f"검색 결과가 {output_file_path}에 저장되었습니다.")
```
