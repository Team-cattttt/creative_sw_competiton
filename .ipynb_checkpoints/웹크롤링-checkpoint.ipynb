```python
import pandas as pd
import selenium
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
import time

# CSV 파일 읽기
data = pd.read_csv("book_60.csv", encoding = 'utf-8')

# 첫 번째 열의 내용 가져오기
queries = data.iloc[:, 0].tolist()

# 크롬 드라이버 경로 설정 및 웹 브라우저 실행
path = "C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회/chromedriver-win64/chromedriver.exe"
service = Service(path)
driver = webdriver.Chrome(service=service)

# 검색 결과를 저장할 리스트
results = []

# Yes24 웹사이트 열기
driver.get("https://www.yes24.com/main/default.aspx")
time.sleep(2)

# 각 검색어에 대해 검색 수행
for query in queries:
    search_box = driver.find_element(By.ID, "query")
    search_box.clear()
    search_box.send_keys(query)
    
    search_button = driver.find_element(By.ID, "searchGoBtn")
    search_button.click()
    
    time.sleep(2)
    
    # 검색 결과 중 첫 번째 항목 클릭
    try:
        first_result = driver.find_element(By.XPATH, '//*[@id="yesSchList"]/li[1]/div/div[2]/a')
        first_result.click()
        time.sleep(2)
        
        # 저자 정보 추출
        author = driver.find_element(By.XPATH, '//*[@id="infoset_intro"]/div[2]/div[1]/table/tbody/tr[1]/td')
        author_text = author.text
        
        # 책소개 정보 추출
        book_intro = driver.find_element(By.XPATH, '//*[@id="infoset_intro"]/div[2]/div[1]/table/tbody/tr[3]/td')
        book_intro_text = book_intro.text
        
        # 데이터 프레임에 추가
        results.append({
            '검색어': query,
            '저자': author_text,
            '책소개': book_intro_text
        })
        
        driver.back()
        time.sleep(2)
    except Exception as e:
        print(f"Error processing query '{query}': {e}")
        continue

# 드라이버 종료
driver.quit()

df = pd.DataFrame(results)
df.to_csv("C:/Users/leeju/Desktop/비교과 활동/창의적 SW 경진대회", index=False, encoding='utf-8-sig')

print(f"검색 결과가 파일에 저장되었습니다.")
```
