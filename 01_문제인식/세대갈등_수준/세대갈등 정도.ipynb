```python
# 1. ‘구분’ 행과 ‘연령’ 행만 추출하기
# 2. <19~29세, 30~39세>, <60세 이상, 65세 이상> 이렇게 데이터 병합한 뒤, <청년층, 노년층>으로 칼럼 변경하기 

import pandas as pd

df3 = pd.read_csv("세대갈등 정도.csv", encoding = 'cp949')
df3
```

```python
df3.shape
```

```python
degree = df3.copy()

drop_list = list(range(0,9)) + list(range(11,14))+ list(range(15, len(degree)))
degree = degree.drop(drop_list)

degree
```

```python
degree.index = ['19~29세', '30~39세', '65세 이상']
degree
```

```python
degree = degree.drop(df3.columns[0], axis = 1)
degree = degree.drop(df3.columns[5], axis = 1)
degree
```

```python
degree.info()
```

```python
degree_young = degree.loc[['19~29세', '30~39세']]
young_averages3 = degree_young.mean()
young = pd.DataFrame([young_averages3], index = ['청년'], columns = degree.columns)

degree = pd.concat([degree, young])

degree
```

```python
degree = degree.drop(['19~29세'])
degree = degree.drop(['30~39세'])
degree = degree.loc[['청년', '65세 이상']]

degree
```
