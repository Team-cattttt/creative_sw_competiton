```python
import pandas as pd
df2 = pd.read_csv("사회갈등의 정도 1+2순위1.csv", encoding = 'cp949')
df2
```

```python
df2.shape
```

```python
conflict2 = df2.copy()

drop_list2 = list(range(0,9)) + list(range(11,14)) + list(range(15, len(conflict2)))
conflict2 = conflict2.drop(drop_list2)

conflict2
```

```python
conflict2.index = ['19~29세', '30~39세', '65세 이상']
conflict2
```

```python
conflict2 = conflict2.drop(df2.columns[0], axis = 1)
conflict2
```

```python
conflict2.info()
```

```python
conflict2 = conflict2.drop('기타', axis=1)
```

```python
conflict2.info()
```

```python
conflict_young2 = conflict2.loc[['19~29세', '30~39세']]

# 행 평균을 계산
young_averages2 = conflict_young2.mean()

# '청년' 데이터프레임 생성
young_df2 = pd.DataFrame([young_averages2], index=['청년'], columns=conflict2.columns)

# 원본 데이터프레임에 '청년' 행 추가
conflict2= pd.concat([conflict2, young_df2])

conflict2
```

```python
conflict2 = conflict2.drop(['19~29세'])
conflict2 = conflict2.drop(['30~39세'])
conflict2
```

```python
conflict2 = conflict2.loc[['청년', '65세 이상']]
conflict2
```

```python
#1순위 데이터 그래프와 형태를 맞추기 위해 순서 변경
conflict2 = conflict2.loc[:, [
    '개인,집단 간 상호이해 부족',
    '개인, 집단 간 가치관 차이',  
    '이해당사자들의 각자 이익 추구',  
    '빈부격차',
    '권력 집중',
    '기회의 불평등 (교육, 취업 등)',
    '갈등 조정 기구 및 제도의 미비'
]]
conflict2
```

```python
import matplotlib.pyplot as plt
data2 = conflict2.loc[['청년', '65세 이상']].mean()
plt.rc('font', family='Malgun Gothic')
colors = ['aqua', 'pink', 'yellow', 'lightgreen', 'skyblue', 'lavender', 'gray']
explode = [0.03] * 7
plt.figure(figsize=(8, 6))
plt.title('사회갈등의 정도 1+2순위', size=18, weight = 'bold')
plt.pie(data2, labels = data2.index, autopct='%.1f%%', colors=colors, explode=explode)
# 그래프 표시
plt.show()
```
