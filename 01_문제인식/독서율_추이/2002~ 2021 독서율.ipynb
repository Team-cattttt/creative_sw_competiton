```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
   'Year': [2002, 2004, 2006, 2007, 2008, 2009, 2010, 2011, 2013, 2015, 2017, 2019, 2021],
    'Reading Rate': [72, 76.3, 75.9, 76.7, 72.2, 71.7, 65.4, 66.8, 72.2, 67.4, 62.3, 55.7, 47.5]
}



reading = pd.DataFrame(data)
plt.rc('font', family='Malgun Gothic')
plt.figure(figsize=(10, 6))

# 데이터프레임에서 연도와 독서율 데이터 추출
years = reading['Year']
values = reading['Reading Rate']

# 꺾은선 그래프 그리기
plt.plot(years, values, marker='s', color='skyblue', linestyle='-', linewidth=2, markersize=8, label='성인')

# 그래프 제목 및 축 라벨 설정
plt.title('독서율 추이')
# plt.xlabel('')
# plt.ylabel('Reading Rate')

# y축 범위 설정
plt.ylim(35, 85)

plt.xticks(years, rotation=45)
plt.grid(True)

# 범례 추가
plt.legend()

# 그래프 보여주기
plt.show()
```
