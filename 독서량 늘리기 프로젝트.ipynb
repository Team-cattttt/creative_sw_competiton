### 1. 데이터 불러오기 및 필요한 데이터만 추출

```python
import pandas as pd
df = pd.read_csv("SPOTFIRE_TEMPLATE_1_2_NEW.csv", encoding = 'cp949')
df
```

```python
df.shape
```

이 데이터는 19개의 행과 17개의 열로 이루어져있다.
그러나 이 데이터에서 우리는 *도서관 이용률*만 필요로 하므로, 필요 없는 행은 삭제해야 한다.

```python
df = df.drop(df.columns[2:19:3], axis = 1)
df
```

사례수 열을 삭제했다.

```python
df.shape
```

```python
df = df.drop(df.columns[3:12:2], axis = 1)
df
```

도서관 비이용 열을 삭제했다.

```python
df = df.drop(0) 
df
```

```python
df.shape # 최종 19행, 7열
```

그러나 14행부터는 따로 분석할 데이터이므로, 데이터를 분할하도록 하겠다.

```python
age = pd.DataFrame(df[4:9]) # 연령 별 도서관 이용률 데이터
age
```

```python
age = age.drop(df.columns[0:2], axis = 1)
age
```

```python
age.index = ['19~29세', '30~39세', '40~49세', '50~59세', '60세 이상']
age
```

인덱스를 연령대로 바꾸었다.

```python
age = age.rename(columns={'2021.1' : '2021','2019.1' : '2019', '2017.1' : '2017','2015.1' : '2015', '2013.1' : '2013'})
age
```

칼럼명을 간단하게 수정하였다.

```python
age = age[['2013', '2015', '2017', '2019', '2021']]
age
```

칼럼의 정렬을 오름차순으로 바꾸었다.

```python
age = age.T # 행 열 전환
age
```

데이터의 행과 열이 바뀌어 있어서, 행과 열을 전환하였다.

```python
vol = pd.DataFrame(df[14:]) # 연간 독서량 데이터
vol
```

데이터를 두 종류로 분류하여 각각 age, vol에 저장하였다.

### 2. 결측값 처리

이제 age 데이터 존재하는 결측치를 확인해보겠다.

```python
pd.isna(age)
```

```python
pd.isna(age).sum() # 결측치 빈도 확인
```

19~29세 행에 결측치가 존재한다. 평균 값으로 이를 대체하도록 하겠다.

```python
age = age.apply(pd.to_numeric, errors='coerce').fillna(0)
```

```python
age['19~29세'].mean()
```

```python
age
```

```python
age['19~29세'] = age['19~29세'].replace(0.0, 26)
age
```

데이터 전처리 끝!
