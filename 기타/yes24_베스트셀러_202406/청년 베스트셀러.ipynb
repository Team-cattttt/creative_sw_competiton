```python
import pandas as pd
```

```python
# 20대 베스트셀러
df1 = pd.read_excel('청년 베스트셀러.xlsx', sheet_name = 0 )
df1
```

```python
df1.shape
```

```python
# '관리분류' 열에서 '자격증'이 포함된 행 제거
df1 = df1[~df1['관리분류'].str.contains('자격증', na=False)]
# '상품명' 열에서 수험서 제거
test = ['토익', 'ETS', '이기적', '해커스', '모의고사', '실기']
pattern = '|'.join(test) 
df1 = df1[~df1['상품명'].str.contains(pattern, na=False, regex=True)]

# 리틀 라이프가 시리즈물이므로, 2편은 제거하였다.
df1 = df1[~df1['상품명'].str.contains('리틀 라이프 2', na=False)]

# 결과 확인
df1
```

```python
df1.shape
```

```python
df1_rank = df1.head(20)
df1_rank
```

```python
df1_rank.loc[:, '순위'] = range(1, 21)
df1_rank.index = range(0, 20)

# 20대 베스트셀러 1~10위
df1_rank
```

```python
# 30대 베스트셀러
df2 = pd.read_excel('청년 베스트셀러.xlsx', sheet_name = 1)
df2
```

```python
# '관리분류' 열에서 '자격증'이 포함된 행 제거
df2 = df2[~df2['관리분류'].str.contains('자격증', na=False)]

# '상품명' 열에서 수험서 제거
test = ['토익', 'ETS', '이기적', '해커스', '모의고사', '실기']
pattern = '|'.join(test) 
df2 = df2[~df2['상품명'].str.contains(pattern, na=False, regex=True)]

# 리틀 라이프가 시리즈물이므로, 2편은 제거하였다.
df2 = df2[~df2['상품명'].str.contains('리틀 라이프 2', na=False)]

# 결과 확인
df2
```

```python
df2.shape
```

```python
df2_rank = df2.head(20)
df2_rank
```

```python
df2_rank.loc[:, '순위'] = range(1, 21)
df2_rank.index = range(0, 20)

# 30대 베스트셀러 1~10위
df2_rank
```

```python
# 빈 리스트에 교차로 행을 추가
young = []

for i in range(len(df1_rank)):
    young.append(df1_rank.iloc[i])
    young.append(df2_rank.iloc[i])

# 리스트를 데이터프레임으로 변환
young = pd.DataFrame(young).reset_index(drop=True)
young
```

```python
young = young.drop(young.columns[4:], axis = 1)
young = young.drop(young.columns[1:3], axis = 1)
young
```

```python
young.shape
```

```python
young = young.drop_duplicates(['상품명'])
young
```

```python
young.shape
```

```python
young = young.head(20)
young
```

```python
young.loc[:, '순위'] = range(1, 21)
young.index = range(0, 20)

# 청년 베스트셀러 1~20위
young
```
