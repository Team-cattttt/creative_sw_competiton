```python
# 20대 베스트셀러
import pandas as pd

df1 = pd.read_excel('청년 베스트셀러.xlsx', sheet_name = 0 )
df1
```

```python
df1.shape
```

```python
# '관리분류' 열에서 '자격증'이 포함된 행 제거
df1 = df1[~df1['관리분류'].str.contains('자격증', na=False)]
# '상품명' 열에서 수험서 제거
test = ['토익', 'ETS', '이기적', '해커스', '모의고사', '실기']
pattern = '|'.join(test) 
df1 = df1[~df1['상품명'].str.contains(pattern, na=False, regex=True)]

# 리틀 라이프가 시리즈물이므로, 2편은 제거하였다.
df1 = df1[~df1['상품명'].str.contains('리틀 라이프 2', na=False)]

# 결과 확인
df1
```

```python
df1.shape
```

```python
df1 = df1.drop(df1.columns[4:], axis = 1)
df1 = df1.drop(df1.columns[1:3], axis = 1)
df1
```

```python
df1.shape
```

```python
best_rank = df1.head(10)
best_rank
```

```python
best_rank.loc[:, '순위'] = range(1, 11)
best_rank.index = range(0, 10)

# 20대 베스트셀러 1~10위
best_rank
```
